# Sistem Fuzzy untuk Menentukan Jumlah Dana Darurat

## Pendahuluan

### Definisi Dana Darurat
Dana darurat adalah sejumlah uang yang disisihkan khusus untuk menghadapi situasi darurat atau tidak terduga, seperti kehilangan pekerjaan, sakit mendadak, atau kerusakan kendaraan. Dana ini berfungsi sebagai jaring pengaman finansial yang memberikan rasa aman dan stabilitas.

### Standar Umum Dana Darurat
Berdasarkan panduan perencanaan keuangan konvensional:
- **Lajang / Single**: 3-6 bulan pengeluaran
- **Menikah / Berkeluarga**: 6-12 bulan pengeluaran
- **Dengan tanggungan banyak**: 9-12 bulan pengeluaran

### Kelemahan Pendekatan Konvensional
1. **Tidak fleksibel**: Menggunakan kategori kaku yang tidak mempertimbangkan variasi kondisi individu
2. **Terlalu sederhana**: Hanya mempertimbangkan status pernikahan, mengabaikan faktor penghasilan dan stabilitas
3. **Sulit diterapkan**: Kondisi di tengah spektrum (misal: penghasilan sedang, tanggungan sedang) sulit dikategorikan
4. **Tidak adaptif**: Tidak dapat menangkap nuansa kompleksitas situasi keuangan yang beragam

### Mengapa Menggunakan Logika Fuzzy?
Logika fuzzy sangat cocok untuk masalah ini karena:
1. **Mengatasi ketidakpastian**: Kondisi keuangan tidak selalu hitam-putih (rendah/tinggi), tetapi sering berada di area abu-abu
2. **Fleksibilitas**: Mampu menangani input yang bersifat linguistik dan gradual ("cukup tinggi", "agak besar")
3. **Pendekatan human-like**: Meniru cara berpikir ahli keuangan yang mempertimbangkan berbagai faktor secara bersamaan
4. **Adaptif terhadap kompleksitas**: Dapat mengintegrasikan multiple variabel (penghasilan, pengeluaran, tanggungan) untuk memberikan rekomendasi yang lebih personal

Sistem ini akan menghasilkan rekomendasi dana darurat dalam **Rupiah** dan **jumlah bulan pengeluaran**, memberikan panduan yang lebih akurat dan dapat disesuaikan dengan kondisi individu.
